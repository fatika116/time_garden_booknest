```python
FILENAME = "books.txt"
DEFAULT_BOOKS = [
    {"id": "01", "title": "The Book Thief", "author": "Markus Zusak", "price": 3500.0, "synopsis": "A young foster girl in Nazi Germany who learns to read and steals books to comfort herself and others during World War II.", "stock": 15},
    {"id": "02", "title": "The Secret Garden", "author": "Frances Hodgson Burnett", "price": 2800.0, "synopsis": "An orphaned girl discovers a hidden, neglected garden.", "stock": 8}
]
books = []
try:
    with open(FILENAME, "r") as file:
        lines = file.readlines()
        for line in lines:
            # Format in file: id|title|author|price|synopsis|stock
            parts = line.strip().split("|")
            if len(parts) == 6:
                books.append({
                    "id": parts[0],
                    "title": parts[1],
                    "author": parts[2],
                    "price": float(parts[3]),
                    "synopsis": parts[4],
                    "stock": int(parts[5])
                })
except FileNotFoundError:
    pass
if not books:
    books = DEFAULT_BOOKS
    try:
        with open(FILENAME, "w") as file:
            for b in books:
                file.write(f"{b['id']}|{b['title']}|{b['author']}|{b['price']}|{b['synopsis']}|{b['stock']}\n")
    except Exception as e:
        print("Error initializing file:", e)
def save_books():
    """Saves all books with their updated IDs, stock, and info to the text file"""
    try:
        with open(FILENAME, "w") as file:
            for b in books:
                file.write(f"{b['id']}|{b['title']}|{b['author']}|{b['price']}|{b['synopsis']}|{b['stock']}\n")
    except Exception as e:
        print("Error saving file:", e)
def view_and_buy():
    print("\n--- Available Books ---")
    for b in books:
        if b['stock'] > 0:
            stock_display = f"PKR {b['price']} | Stock available: {b['stock']}"
        else:
            stock_display = f"PKR {b['price']} | (Out of Stock)"   
        print(f"ID [{b['id']}]: {b['title']} by {b['author']} — {stock_display}")     
    book_id = input("\nEnter the Book ID to read synopsis & buy: ").strip()
    selected = None
    for b in books:
        if b['id'] == book_id:
            selected = b
            break      
    if selected:
        print(f"\n--- {selected['title']} ---")
        print(f"Author: {selected['author']}")
        print(f"Price: PKR {selected['price']}")
        print(f"Synopsis: {selected['synopsis']}")
        if selected['stock'] <= 0:
            print("\nSorry, this book is currently out of stock and cannot be purchased.")
            return
        buy = input("\nDo you want to buy this book? (yes/no): ").strip().lower()
        if buy == "yes":
            selected['stock'] -= 1  
            save_books()  
            print(f"Purchase successful! You bought '{selected['title']}'. Remaining stock: {selected['stock']}")
        else:
            print("Returned to menu.")
    else:
        print("Invalid Book ID. Please try again.")
def add_book():
    print("\n--- Add New Book (Admin) ---")
    book_id = input("Enter a unique book ID:").strip()
    for b in books:
        if b['id'] == book_id:
            print("Error: A book with this ID already exists! Please use a different ID.")
            return
    title = input("Enter book title: ").strip()
    author = input("Enter author name: ").strip()
    try:
        price = float(input("Enter price in PKR: "))
        stock = int(input("Enter initial stock quantity: "))
        synopsis = input("Enter book synopsis: ").strip()
        books.append({
            "id": book_id,
            "title": title,
            "author": author,
            "price": price,
            "synopsis": synopsis,
            "stock": stock
        })
        save_books()
        print(f"'{title}' added and saved successfully with ID [{book_id}]!")
    except ValueError:
        print("Error: Price must be a decimal/number and stock must be a whole number.")
def booknest_portal():
    print("=== Welcome to Time Garden Booknest ===")
    while True:
        print("\n1. Browse & Buy Books")
        print("2. Add Book (Admin)")
        print("3. Exit")
        choice = input("Enter your choice: ").strip()
        if choice == "1":
            view_and_buy()
        elif choice == "2":
            add_book()
        elif choice == "3":
            print("Thank you for visiting Time Garden Booknest. Goodbye!")
            break
        else:
            print("Invalid choice, please try again.")
booknest_portal()
```
