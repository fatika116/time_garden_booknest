```python
## BOOKS_FILE = "books.txt"
ORDERS_FILE = "orders.txt"
ADMIN_PIN = "hahahehehoho"
DEFAULT_BOOKS = [
    {
        "id": "01",
        "title": "The Housemaid",
        "author": "Freida McFadden",
        "genre": "Thriller",
        "price": 2400.0,
        "synopsis": "A struggling young woman takes a live-in housekeeping job for a wealthy couple, only to discover dark, dangerous secrets locked behind closed doors.",
        "stock": 12
    },
    {
        "id": "02",
        "title": "A Little Life",
        "author": "Hanya Yanagihara",
        "genre": "Contemporary Fiction",
        "price": 3800.0,
        "synopsis": "Four college friends move to New York City to make their way, bound together by devotion to Jude, a brilliant man haunted by an unspeakable past.",
        "stock": 7
    },
    {
        "id": "03",
        "title": "And Then There Were None",
        "author": "Agatha Christie",
        "genre": "Mystery",
        "price": 2100.0,
        "synopsis": "Ten strangers are lured to an isolated island mansion off the Devon coast, where they are accused of past crimes and murdered one by one.",
        "stock": 14
    },
    {
        "id": "04",
        "title": "The Book Thief",
        "author": "Markus Zusak",
        "genre": "Historical Fiction",
        "price": 3500.0,
        "synopsis": "A young foster girl in Nazi Germany who learns to read and steals books to comfort herself and others during World War II.",
        "stock": 15
    },
    {
        "id": "05",
        "title": "I'm Glad My Mom Died",
        "author": "Jennette McCurdy",
        "genre": "Memoir",
        "price": 3200.0,
        "synopsis": "A heartbreaking and darkly hilarious memoir by a former child star recounting her struggles with eating disorders, addiction, and a domineering mother.",
        "stock": 9
    },
    {
        "id": "06",
        "title": "The Alchemist",
        "author": "Paulo Coelho",
        "genre": "Philosophy & Adventure",
        "price": 1950.0,
        "synopsis": "A young Andalusian shepherd travels from Spain to the Egyptian desert in search of a buried treasure, discovering his Personal Legend along the way.",
        "stock": 20
    },
    {
        "id": "07",
        "title": "Gone Girl",
        "author": "Gillian Flynn",
        "genre": "Thriller",
        "price": 2600.0,
        "synopsis": "When a woman suddenly disappears on her fifth wedding anniversary, mounting media pressure and disturbing clues turn suspicion onto her husband.",
        "stock": 11
    },
    {
        "id": "08",
        "title": "And the Mountains Echoed",
        "author": "Khaled Hosseini",
        "genre": "Historical Fiction",
        "price": 3100.0,
        "synopsis": "Spanning generations and continents from Kabul to Paris and San Francisco, a heart-wrenching decision separates a brother and sister and ripples through lives.",
        "stock": 10
    },
    {
        "id": "09",
        "title": "1984",
        "author": "George Orwell",
        "genre": "Dystopian Sci-Fi",
        "price": 1800.0,
        "synopsis": "In a totalitarian superstate ruled by Big Brother, a low-ranking Party member risks everything by falling in love and questioning the regime's rewritten truth.",
        "stock": 18
    },
    {
        "id": "10",
        "title": "If Tomorrow Comes",
        "author": "Sidney Sheldon",
        "genre": "Suspense",
        "price": 2300.0,
        "synopsis": "Framed for a crime she did not commit and robbed of everything she loved, a brilliant woman reinvents herself as an international con artist seeking vengeance.",
        "stock": 6
    },
    {
        "id": "11",
        "title": "Babel",
        "author": "R. F. Kuang",
        "genre": "Fantasy",
        "price": 3900.0,
        "synopsis": "In 1830s Oxford, a Canton-born orphan enters the Royal Institute of Translation, where silver-working powers the British Empire—forcing him to choose between loyalty and revolution.",
        "stock": 8
    },
    {
        "id": "12",
        "title": "The Secret Garden",
        "author": "Frances Hodgson Burnett",
        "genre": "Classic",
        "price": 2800.0,
        "synopsis": "An orphaned girl discovers a hidden, neglected garden.",
        "stock": 8
    },
    {
        "id": "13",
        "title": "A Thousand Splendid Suns",
        "author": "Khaled Hosseini",
        "genre": "Historical Fiction",
        "price": 2950.0,
        "synopsis": "Two Afghan women from vastly different backgrounds are brought together by war, loss, and a tyrannical husband, forging an unbreakable maternal bond.",
        "stock": 13
    },
    {
        "id": "14",
        "title": "Tell Me Your Dreams",
        "author": "Sidney Sheldon",
        "genre": "Mystery",
        "price": 2250.0,
        "synopsis": "Three young women are stalked by a suspected killer, leading to a shocking arrest and a gripping courtroom battle over dissociative identity disorder.",
        "stock": 9
    }
]
books = []
orders = []
cart = []
# -------------------- FILE HANDLING --------------------
def load_books():
    global books
    books = []
    try:
        with open(BOOKS_FILE, "r") as file:
            for line in file:
                parts = line.strip().split("|")
                if len(parts) == 7:
                    books.append({
                        "id": parts[0],
                        "title": parts[1],
                        "author": parts[2],
                        "genre": parts[3],
                        "price": float(parts[4]),
                        "synopsis": parts[5],
                        "stock": int(parts[6])
                    })
    except FileNotFoundError:
        pass
    if not books:
        books = [b.copy() for b in DEFAULT_BOOKS]
        save_books()
def save_books():
    try:
        with open(BOOKS_FILE, "w") as file:
            for b in books:
                file.write(
                    f"{b['id']}|{b['title']}|{b['author']}|{b['genre']}|"
                    f"{b['price']}|{b['synopsis']}|{b['stock']}\n"
                )
    except Exception as e:
        print("Error saving books file:", e)
def load_orders():
    global orders
    orders = []
    try:
        with open(ORDERS_FILE, "r") as file:
            for line in file:
                parts = line.strip().split("|")
                if len(parts) == 5:
                    orders.append({
                        "order_id": parts[0],
                        "customer": parts[1],
                        "items": parts[2],
                        "total": float(parts[3]),
                        "status": parts[4]
                    })
    except FileNotFoundError:
        pass
def save_orders():
    try:
        with open(ORDERS_FILE, "w") as file:
            for o in orders:
                file.write(
                    f"{o['order_id']}|{o['customer']}|{o['items']}|"
                    f"{o['total']}|{o['status']}\n"
                )
    except Exception as e:
        print("Error saving orders file:", e)
# -------------------- HELPER FUNCTIONS --------------------
def find_book_by_id(book_id):
    for b in books:
        if b["id"] == book_id:
            return b
    return None
def display_book_list(book_list, heading="Available Books"):
    print(f"\n--- {heading} ---")
    if not book_list:
        print("No books found.")
        return
    for b in book_list:
        if b["stock"] > 0:
            stock_display = f"PKR {b['price']} | Stock: {b['stock']}"
        else:
            stock_display = f"PKR {b['price']} | (Out of Stock)"
        print(f"ID [{b['id']}]: {b['title']} by {b['author']} [{b['genre']}] — {stock_display}")
def inspect_and_add_to_cart():
    book_id = input("\nEnter a Book ID to view details & add to cart (or press Enter to go back): ").strip()
    if not book_id:
        return
    selected = find_book_by_id(book_id)
    if not selected:
        print("Invalid Book ID.")
        return
    print(f"\n--- {selected['title']} ---")
    print(f"Author:{selected['author']}")
    print(f"Genre:{selected['genre']}")
    print(f"Price:PKR {selected['price']}")
    print(f"Stock:{selected['stock']}")
    print(f"Synopsis:{selected['synopsis']}")
    if selected["stock"] <= 0:
        print("\nSorry, this book is currently out of stock.")
        return
    in_cart = 0
    for item in cart:
        if item["id"] == selected["id"]:
            in_cart = item["qty"]
            break
    available_to_add = selected["stock"] - in_cart
    if available_to_add <= 0:
        print("\nYou already have all available copies of this book in your cart.")
        return
    choice = input("\nAdd this book to your shopping cart? (yes/no): ").strip().lower()
    if choice == "yes":
        try:
            qty = int(input(f"Enter quantity (1 - {available_to_add}): ").strip())
            if qty < 1 or qty > available_to_add:
                print("Invalid quantity.")
                return
            for item in cart:
                if item["id"] == selected["id"]:
                    item["qty"] += qty
                    break
            else:
                cart.append({
                    "id": selected["id"],
                    "title": selected["title"],
                    "price": selected["price"],
                    "qty": qty
                })
            print(f"Added {qty} x '{selected['title']}' to your cart!")
        except ValueError:
            print("Please enter a valid whole number.")
# -------------------- CUSTOMER FEATURES --------------------
def browse_books():
    display_book_list(books, "All Available Books")
    inspect_and_add_to_cart()
def search_books():
    print("\n--- Search Books ---")
    keyword = input("Enter title or author name to search: ").strip().lower()
    if not keyword:
        print("Search term cannot be empty.")
        return
    matches = [
        b for b in books
        if keyword in b["title"].lower() or keyword in b["author"].lower()
    ]
    display_book_list(matches, f"Search Results for '{keyword}'")
    if matches:
        inspect_and_add_to_cart()
def explore_genres():
    print("\n--- Explore by Genre ---")
    genres = sorted(list({b["genre"] for b in books}))
    for idx, g in enumerate(genres, start=1):
        count = sum(1 for b in books if b["genre"] == g)
        print(f"{idx}. {g} ({count} book{'s' if count != 1 else ''})")
    choice = input("\nSelect a genre number (or press Enter to go back): ").strip()
    if not choice:
        return
    try:
        index = int(choice) - 1
        if 0 <= index < len(genres):
            chosen_genre = genres[index]
            filtered = [b for b in books if b["genre"] == chosen_genre]
            display_book_list(filtered, f"Genre: {chosen_genre}")
            inspect_and_add_to_cart()
        else:
            print("Invalid genre number.")
    except ValueError:
        print("Please enter a valid number.")
def view_cart_and_checkout():
    while True:
        print("\n--- Your Shopping Cart ---")
        if not cart:
            print("Your cart is currently empty.")
            return
        total = 0.0
        for idx, item in enumerate(cart, start=1):
            subtotal = item["price"] * item["qty"]
            total += subtotal
            print(f"{idx}. [{item['id']}] {item['title']} — PKR {item['price']} x {item['qty']} = PKR {subtotal}")
        print(f"\nTotal Amount: PKR {total}")
        print("\n1. Place Order (Checkout)")
        print("2. Remove an Item from Cart")
        print("3. Clear Cart")
        print("4. Back to Store")
        action = input("Choose an option: ").strip()
        if action == "1":
            customer_name = input("Enter your full name for the order: ").strip().replace("|", "")
            if not customer_name:
                print("Name cannot be empty.")
                continue
            for item in cart:
                book = find_book_by_id(item["id"])
                if not book or book["stock"] < item["qty"]:
                    print(f"Sorry, '{item['title']}' is out of stock.")
                    return
            summary_parts = []
            for item in cart:
                book = find_book_by_id(item["id"])
                book["stock"] -= item["qty"]
                summary_parts.append(f"{item['id']}:{item['title']} (x{item['qty']})")
            order_id = f"ORD-{len(orders) + 1:03d}"
            orders.append({
                "order_id": order_id,
                "customer": customer_name,
                "items": "; ".join(summary_parts),
                "total": total,
                "status": "Confirmed"
            })
            save_books()
            save_orders()
            cart.clear()
            print(f"\nOrder placed successfully! Your Order ID is [{order_id}].")
            return
        elif action == "2":
            try:
                num = int(input("Enter item number to remove: ").strip()) - 1
                if 0 <= num < len(cart):
                    removed = cart.pop(num)
                    print(f"Removed '{removed['title']}' from cart.")
                else:
                    print("Invalid item number.")
            except ValueError:
                print("Please enter a valid number.")
        elif action == "3":
            cart.clear()
            print("Cart cleared.")
            return
        elif action == "4":
            return
        else:
            print("Invalid option.")
def manage_orders():
    print("\n--- Place & Manage Orders ---")
    if not orders:
        print("No orders have been placed yet.")
        return
    for o in orders:
        print(f"[{o['order_id']}] Customer: {o['customer']} | Total: PKR {o['total']} | Status: {o['status']}")
        print(f"   Items: {o['items']}")
    order_id = input("\nEnter an Order ID to cancel it (or press Enter to go back): ").strip().upper()
    if not order_id:
        return
    for o in orders:
        if o["order_id"] == order_id:
            if o["status"] == "Cancelled":
                print("This order is already cancelled.")
                return
            confirm = input(f"Are you sure you want to cancel {order_id}? (yes/no): ").strip().lower()
            if confirm == "yes":
                # Restore stock from order summary ("id:title (xqty); ...")
                for entry in o["items"].split("; "):
                    try:
                        book_id_part, rest = entry.split(":", 1)
                        qty_str = rest.rsplit("(x", 1)[1].rstrip(")")
                        qty = int(qty_str)
                        book = find_book_by_id(book_id_part)
                        if book:
                            book["stock"] += qty
                    except Exception:
                        pass
                o["status"] = "Cancelled"
                save_books()
                save_orders()
                print(f"Order [{order_id}] has been cancelled and stock restored.")
            return
    print("Order ID not found.")
# -------------------- HIDDEN ADMIN FEATURES --------------------
def add_book():
    print("\n--- Add New Book ---")
    book_id = input("Enter a unique book ID: ").strip().replace("|", "")
    if not book_id:
        print("Book ID cannot be empty.")
        return
    if find_book_by_id(book_id):
        print("Error: A book with this ID already exists!")
        return
    title = input("Enter book title: ").strip().replace("|", "")
    author = input("Enter author name: ").strip().replace("|", "")
    genre = input("Enter book genre: ").strip().replace("|", "")
    if not title or not author or not genre:
        print("Title, author, and genre cannot be empty.")
        return
    try:
        price = float(input("Enter price in PKR: ").strip())
        stock = int(input("Enter initial stock quantity: ").strip())
        if price < 0 or stock < 0:
            print("Price and stock cannot be negative.")
            return
        synopsis = input("Enter book synopsis: ").strip().replace("|", "")
        books.append({
            "id": book_id,
            "title": title,
            "author": author,
            "genre": genre,
            "price": price,
            "synopsis": synopsis,
            "stock": stock
        })
        save_books()
        print(f"'{title}' added and saved successfully with ID [{book_id}]!")
    except ValueError:
        print("Error: Price must be a number and stock must be a whole number.")
def restock_book():
    print("\n--- Restock Book ---")
    display_book_list(books, "Current Inventory")
    book_id = input("\nEnter Book ID to restock: ").strip()
    book = find_book_by_id(book_id)
    if not book:
        print("Invalid Book ID.")
        return
    try:
        added = int(input(f"Enter copies to add to '{book['title']}' (Current stock: {book['stock']}): ").strip())
        if added <= 0:
            print("Quantity must be greater than zero.")
            return
        book["stock"] += added
        save_books()
        print(f"Updated stock for '{book['title']}': {book['stock']}")
    except ValueError:
        print("Please enter a valid whole number.")
def sales_and_inventory_report():
    print("\n=== Sales & Inventory Report ===")
    confirmed_orders = [o for o in orders if o["status"] == "Confirmed"]
    cancelled_orders = [o for o in orders if o["status"] == "Cancelled"]
    total_revenue = sum(o["total"] for o in confirmed_orders)
    print(f"Total Confirmed Orders: {len(confirmed_orders)}")
    print(f"Total Cancelled Orders: {len(cancelled_orders)}")
    print(f"Total Revenue:          PKR {total_revenue}")
    total_titles = len(books)
    total_units = sum(b["stock"] for b in books)
    inventory_value = sum(b["price"] * b["stock"] for b in books)
    low_stock = [b for b in books if b["stock"] <= 5]
    print(f"\nTotal Book Titles:      {total_titles}")
    print(f"Total Units in Stock:   {total_units}")
    print(f"Total Inventory Value:  PKR {inventory_value}")
    print("\nLow Stock Alerts (5 or fewer left):")
    if low_stock:
        for b in low_stock:
            print(f" - [{b['id']}] {b['title']}: {b['stock']} left")
    else:
        print(" - All books are well stocked!")
def admin_panel():
    pin = input("Enter Admin PIN: ").strip()
    if pin != ADMIN_PIN:
        print("Incorrect PIN. Returning to store.")
        return
    while True:
        print("\n=== Admin Panel ===")
        print("1. Add New Book")
        print("2. Restock Existing Book")
        print("3. Sales & Inventory Report")
        print("4. Return to Storefront")
        choice = input("Enter your choice: ").strip()
        if choice == "1":
            add_book()
        elif choice == "2":
            restock_book()
        elif choice == "3":
            sales_and_inventory_report()
        elif choice == "4":
            break
        else:
            print("Invalid choice, please try again.")
# -------------------- MAIN STOREFRONT --------------------
def booknest_portal():
    load_books()
    load_orders()
    print("=== Welcome to Time Garden Booknest ===")
    while True:
        cart_count = sum(item["qty"] for item in cart)
        print("\n1. Browse Available Books")
        print("2. Search Books (by Title or Author)")
        print("3. Explore Book Genres")
        print(f"4. Shopping Cart & Checkout ({cart_count} item{'s' if cart_count != 1 else ''})")
        print("5. View & Manage Orders")
        print("6. Exit")
        choice = input("Enter your choice: ").strip().lower()
        if choice == "1":
            browse_books()
        elif choice == "2":
            search_books()
        elif choice == "3":
            explore_genres()
        elif choice == "4":
            view_cart_and_checkout()
        elif choice == "5":
            manage_orders()
        elif choice == "6":
            print("Thank you for visiting Time Garden Booknest. Goodbye!")
            break
        elif choice == "admin":
            admin_panel()
        else:
            print("Invalid choice, please try again.")
booknest_portal()
```
