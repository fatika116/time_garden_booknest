```python
BOOKS_FILE = "books.txt"
ORDERS_FILE = "orders.txt"
ADMIN_PIN = "hahahehehoho"
DEFAULT_BOOKS = [
    {"id": "01", "title": "The Housemaid", "author": "Freida McFadden", "genre": "Thriller", "price": 2400.0, "synopsis": "A young woman takes a live-in housekeeping job for a wealthy couple, uncovering dark secrets.", "stock": 12},
    {"id": "02", "title": "A Little Life", "author": "Hanya Yanagihara", "genre": "Fiction", "price": 3800.0, "synopsis": "Four college friends move to New York City, bound by devotion to Jude, a man with a painful past.", "stock": 7},
    {"id": "03", "title": "And Then There Were None", "author": "Agatha Christie", "genre": "Mystery", "price": 2100.0, "synopsis": "Ten strangers are lured to an isolated island mansion and murdered one by one.", "stock": 14},
    {"id": "04", "title": "The Book Thief", "author": "Markus Zusak", "genre": "Historical", "price": 3500.0, "synopsis": "A foster girl in Nazi Germany steals books to comfort herself and others during WWII.", "stock": 15},
    {"id": "05", "title": "I'm Glad My Mom Died", "author": "Jennette McCurdy", "genre": "Memoir", "price": 3200.0, "synopsis": "A former child star recounts her struggles with eating disorders, addiction, and a domineering mother.", "stock": 9},
    {"id": "06", "title": "The Alchemist", "author": "Paulo Coelho", "genre": "Philosophy", "price": 1950.0, "synopsis": "A shepherd travels from Spain to Egypt in search of treasure and his Personal Legend.", "stock": 20},
    {"id": "07", "title": "Gone Girl", "author": "Gillian Flynn", "genre": "Thriller", "price": 2600.0, "synopsis": "When a woman disappears on her wedding anniversary, clues turn suspicion onto her husband.", "stock": 11},
    {"id": "08", "title": "And the Mountains Echoed", "author": "Khaled Hosseini", "genre": "Historical", "price": 3100.0, "synopsis": "Spanning generations from Kabul to Paris, a difficult choice separates a brother and sister.", "stock": 10},
    {"id": "09", "title": "1984", "author": "George Orwell", "genre": "Sci-Fi", "price": 1800.0, "synopsis": "In a totalitarian state ruled by Big Brother, a man risks everything by questioning the regime.", "stock": 18},
    {"id": "10", "title": "If Tomorrow Comes", "author": "Sidney Sheldon", "genre": "Suspense", "price": 2300.0, "synopsis": "Framed for a crime she didn't commit, a woman becomes an international con artist seeking revenge.", "stock": 6},
    {"id": "11", "title": "Babel", "author": "R. F. Kuang", "genre": "Fantasy", "price": 3900.0, "synopsis": "In 1830s Oxford, an orphan joins a translation institute where silver-working powers the empire.", "stock": 8},
    {"id": "12", "title": "The Secret Garden", "author": "Frances Hodgson Burnett", "genre": "Classic", "price": 2800.0, "synopsis": "An orphaned girl discovers a hidden, neglected garden.", "stock": 8},
    {"id": "13", "title": "A Thousand Splendid Suns", "author": "Khaled Hosseini", "genre": "Historical", "price": 2950.0, "synopsis": "Two Afghan women are brought together by war and forge an unbreakable bond.", "stock": 13},
    {"id": "14", "title": "Tell Me Your Dreams", "author": "Sidney Sheldon", "genre": "Mystery", "price": 2250.0, "synopsis": "Three young women are stalked by a killer, leading to a gripping courtroom battle.", "stock": 9}
]
books, orders, cart = [], [], []
try:
    with open(BOOKS_FILE, "r") as f:
        for line in f:
            p = line.strip().split("|")
            if len(p) == 7:
                books.append({"id": p[0], "title": p[1], "author": p[2], "genre": p[3], "price": float(p[4]), "synopsis": p[5], "stock": int(p[6])})
except FileNotFoundError:
    pass
if not books:
    books = DEFAULT_BOOKS
def save_all():
    with open(BOOKS_FILE, "w") as f:
        for b in books:
            f.write(f"{b['id']}|{b['title']}|{b['author']}|{b['genre']}|{b['price']}|{b['synopsis']}|{b['stock']}\n")
    with open(ORDERS_FILE, "w") as f:
        for o in orders:
            f.write(f"{o['id']}|{o['name']}|{o['items']}|{o['total']}\n")
try:
    with open(ORDERS_FILE, "r") as f:
        for line in f:
            p = line.strip().split("|")
            if len(p) == 4:
                orders.append({"id": p[0], "name": p[1], "items": p[2], "total": float(p[3])})
except FileNotFoundError:
    pass
save_all()
def show_and_buy(book_list):
    if not book_list:
        print("No books found.")
        return
    for b in book_list:
        print(f"[{b['id']}] {b['title']} - {b['author']} ({b['genre']}) | PKR {b['price']} | Stock: {b['stock']}")
    bid = input("\nEnter Book ID to view & add to cart (or Enter to skip): ").strip()
    for b in book_list:
        if b["id"] == bid:
            print(f"\n{b['title']} by {b['author']}\nSynopsis: {b['synopsis']}")
            if b["stock"] > 0 and input("Add to cart? (yes/no): ").lower() == "yes":
                cart.append(b)
                print("Added to cart!")
            elif b["stock"] <= 0:
                print("Out of stock!")
            return
print("=== Welcome to Time Garden Booknest ===")
while True:
    print(f"\n1. Browse  2. Search  3. Genres  4. Cart ({len(cart)})  5. Orders  6. Exit")
    ch = input("Choice: ").strip().lower()
    if ch == "1":
        show_and_buy(books)
    elif ch == "2":
        q = input("Search title or author: ").strip().lower()
        show_and_buy([b for b in books if q in b["title"].lower() or q in b["author"].lower()])
    elif ch == "3":
        g = input("Enter genre (Thriller, Fiction, Mystery, Historical, Memoir, Philosophy, Sci-Fi, Suspense, Fantasy, Classic): ").strip().lower()
        show_and_buy([b for b in books if b["genre"].lower() == g])
    elif ch == "4":
        if not cart:
            print("Cart is empty.")
        else:
            total = sum(b["price"] for b in cart)
            for b in cart:
                print(f"- {b['title']} (PKR {b['price']})")
            print(f"Total: PKR {total}")
            if input("Place order? (yes/no): ").lower() == "yes":
                name = input("Your name: ").strip()
                for b in cart:
                    b["stock"] -= 1
                orders.append({"id": f"ORD-{len(orders)+1}", "name": name, "items": ", ".join(b["title"] for b in cart), "total": total})
                cart.clear()
                save_all()
                print("Order placed!")
    elif ch == "5":
        for o in orders:
            print(f"[{o['id']}] {o['name']}: {o['items']} — PKR {o['total']}")
        if not orders:
            print("No orders yet.")
    elif ch == "admin":
        if input("Admin PIN: ") == ADMIN_PIN:
            act = input("1. Add Book  2. Sales & Stock Report: ").strip()
            if act == "1":
                books.append({
                    "id": input("ID: "), "title": input("Title: "), "author": input("Author: "),
                    "genre": input("Genre: "), "price": float(input("Price: ")),
                    "synopsis": input("Synopsis: "), "stock": int(input("Stock: "))
                })
                save_all()
                print("Book added!")
            elif act == "2":
                print(f"Orders: {len(orders)} | Revenue: PKR {sum(o['total'] for o in orders)} | Total Stock: {sum(b['stock'] for b in books)}")
        else:
            print("Wrong PIN!")
    elif ch == "6":
        print("Thank You for visiting Time Garden BookNest. Goodbye!")
        break
```
